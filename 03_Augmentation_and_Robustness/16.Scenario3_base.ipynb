```python
E_GRID = np.arange(-10,60.1,0.5)
```

```python
extra_tree_params = {'n_estimators': 300,
                     'max_features': 0.3, # ~30% точек при каждом сплите
                     'max_depth': 20,
                     'min_samples_leaf': 3,
                     'min_samples_split': 6,
                     'n_jobs': -1,
                     'random_state': 42,
                    }
# Gemini
catboost_params = {
    # --- Основные параметры ---
    'iterations': 2000,          
    'learning_rate': 0.03,       
    'depth': 6,                  
    'loss_function': 'RMSE',
    
    # --- Регуляризация (Борьба с переобучением) ---
    'l2_leaf_reg': 3.0,          # L2-регуляризация (чем выше, тем консервативнее модель)
    'random_strength': 1.0,      # Добавляет случайность при выборе сплитов
    'bagging_temperature': 0.8,  # Интенсивность байесовского бутстрапа (0 - выкл, 1 - макс)
    
    # --- Специфика данных (Спектры) ---
    'bootstrap_type': 'Bayesian', 
    'leaf_estimation_method': 'Newton', # Обычно точнее для регрессии
    'feature_border_type': 'UniformAndQuantiles', # Важно для гладких спектров
    
    # --- Технические параметры ---
    'early_stopping_rounds': 100,
    'verbose': False,
    'allow_writing_files': False,
    'random_seed': 42
}
```

```python
class XanesNN_JanisLike(nn.Module):
    def __init__(self,in_dim):
        super().__init__()
        self.in_dim=in_dim
        self.model = nn.Sequential(nn.Linear(self.in_dim, 387), # Входной слой (129) -> Скрытый 1 
                                   nn.Tanh(), # nn.Tanh(), у Яниса Tanh 
                                   nn.Linear(387, 387), # Скрытый 1 -> Скрытый 2 
                                   nn.Tanh(), # 
                                   nn.Linear(387, 1)
                                  )
    
    def forward(self, x):
        return self.model(x)

class ReshapeWrapper(nn.Module):
    def __init__(self, dim):
        super().__init__()
        self.dim = dim
    def forward(self, x):
        return x.unsqueeze(self.dim)

        
# 2. Обновленная архитектура Torch
class XANES_CNN(nn.Module):
    def __init__(self, in_dim):
        super().__init__()
        self.net = nn.Sequential(
            ReshapeWrapper(1),             # Теперь это корректный nn.Module [Batch, 1, Seq]
            nn.Conv1d(in_channels=1, out_channels=16, kernel_size=5, padding='same'), nn.GELU(),
            nn.MaxPool1d(2),
            nn.Conv1d(in_channels=16, out_channels=32, kernel_size=5, padding='same'), nn.GELU(),
            nn.AdaptiveAvgPool1d(10),       # Сжимаем временную/энергетическую ось до 10
            nn.Flatten(),
            nn.Linear(32 * 10, 64), nn.ReLU(),
            nn.Linear(64, 1)
        )

    def forward(self, x):
        return self.net(x)
```

```python
IN_DIM=141 # Размер входного спектра.
nn_params = {'epochs':480, 
             'batch_size':512, 
             'lr':1e-3, 
             'weight_decay': 0, #0, #1e-3, 
             'loss_params':{'alpha':1.0, 'beta':1., 'gamma':0.}, # MSE (0,1,0)-MAE, (0,0,1)-Mape 
             'T_0':80, 
             'T_mult':2, 
             'eta_min':1e-6, 
             'patience':160, 
             'min_delta':0,
            }
```
